```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
## rules for multiplication validity

m = 4
n = 3
k = 6

# make some matrices
A = np.random.randn(m,n)
B = np.random.randn(n,k)
C = np.random.randn(m,k)

# test which multiplications are valid.
# Think of your answer first, then test.
try:
    print('np.matmul(A,B)', "=", np.matmul(A,B))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(A,A)', "=", np.matmul(A,A))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(A.T,C)', "=", np.matmul(A.T,C))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(B,B.T)', "=", np.matmul(B,B.T))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(B.T,B)', "=", np.matmul(B.T,B))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(B,C)', "=", np.matmul(B,C))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(C,B)', "=", np.matmul(C,B))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(C.T,B)', "=", np.matmul(C.T,B))
except ValueError as error:
    print("Expected incompatible shapes:", error)
try:
    print('np.matmul(C,B.T)', "=", np.matmul(C,B.T))
except ValueError as error:
    print("Expected incompatible shapes:", error)
```
