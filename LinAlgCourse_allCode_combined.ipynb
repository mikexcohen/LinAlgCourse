# VIDEO: Diagonalization

```python
## eigendecomposition of A

# make a symmetric matrix
A = np.round(10*np.random.randn(4,4))
A = A.T@A

# eigendecomposition
evals,evecs = np.linalg.eigh(A)

# test reconstruction
Ap = evecs @ np.diag(evals) @ np.linalg.inv(evecs)


# plot
plt.subplot(121)
plt.imshow(A)
plt.axis('off')
plt.title('A')

plt.subplot(122)
plt.imshow(Ap)
plt.axis('off')
plt.title(r'$V\Lambda V^{-1}$')

plt.show()

# subtract the two (should be zero with rounding errors)
recondiff = A-Ap

# reconstruction error (note the numerical inaccuracies)
rmsA = np.sqrt( np.mean(np.square( np.reshape(recondiff,(1,-1)) )) )

print('Reconstruction RMS:', rmsA)
```

# VIDEO: Eigenvectors of repeated eigenvalues
---


```python
# Draw a plane from a point and normal using public Matplotlib APIs.
def plot_plane(ax, point, normal, size=10, color='g'):
    point = np.asarray(point, dtype=float)
    normal = np.asarray(normal, dtype=float)
    norm = np.linalg.norm(normal)
    if norm == 0:
        raise ValueError('A plane needs a nonzero normal.')
    normal = normal / norm
    # Pick the coordinate axis least parallel to the normal.
    reference = np.eye(3)[np.argmin(np.abs(normal))]
    u = np.cross(normal, reference)
    u /= np.linalg.norm(u)
    v = np.cross(normal, u)
    a, b = np.meshgrid(np.linspace(-size, size, 21),
                       np.linspace(-size, size, 21))
    xyz = point[:, None, None] + u[:, None, None]*a + v[:, None, None]*b
    return ax.plot_surface(*xyz, color=color, alpha=.35)
```

```python
### The plane helper is defined in the preceding cell.

# a matrix
A = [ [ 5,  -1,  0], 
      [-1,   5,  0],
      [1/3,-1/3, 4] ]

# its eigendecomposition
D,V = np.linalg.eig(A)
D, V = np.real_if_close(D), np.real_if_close(V)

# sort eigenvalues
i = np.argsort(D)
D = D[i]
V = V[:,i]


## plot eigenvectors
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# plot eigenvectors of A
ax.plot([0,V[0,0]],[0,V[1,0]],[0,V[2,0]],'r',label=r'v_1 ($\lambda$=%g)' %D[0])
ax.plot([0,V[0,1]],[0,V[1,1]],[0,V[2,1]],'k',label=r'v_2 ($\lambda$=%g)' %D[1])
ax.plot([0,V[0,2]],[0,V[1,2]],[0,V[2,2]],'b',label=r'v_3 ($\lambda$=%g)' %D[2])
plt.legend()

# plot subspace spanned by same-eigenvalued eigenvectors
plot_plane(ax, np.zeros(3), np.cross(V[:,0], V[:,1]), size=3)
ax.set_xlim3d(-1,1)
ax.set_ylim3d(-1,1)
ax.set_zlim3d(-1,1)

plt.show()
```

# VIDEO: Left inverse and right inverse
---


```python
# A tall matrix has a left inverse; its transpose has a right inverse.
m, n = 6, 3
A = np.random.randn(m, n)
AtA = A.T @ A
AAt = A @ A.T
print('Shape/rank of A^T A:', AtA.shape, np.linalg.matrix_rank(AtA))
print('Shape/rank of A A^T:', AAt.shape, np.linalg.matrix_rank(AAt))

Aleft = np.linalg.solve(AtA, A.T)
I_left = Aleft @ A
# A A^T has rank n < m, so it cannot be inverted.
# Use the wide matrix B for the right-inverse example instead.
B = A.T
Aright = np.linalg.solve(B @ B.T, B).T
I_right = B @ Aright
print('Left inverse check (Aleft @ A):', I_left)
print('Right inverse check (B @ Aright):', I_right)
assert np.allclose(I_left, np.eye(n))
assert np.allclose(I_right, np.eye(n))

AtA_inv = np.linalg.inv(AtA)
I_AtA = AtA_inv @ AtA
# A pseudoinverse of the singular Gram matrix gives a projector, not I.
AAt_inv = np.linalg.pinv(AAt)
I_AAt = AAt_inv @ AAt
print('Inverse check for A^T A:', I_AtA)
print('Projector for singular A A^T:', I_AAt)
```

# VIDEO: Least-squares example 1
---


```python
# data
data = np.array([[-4,0,-3,1,2,8,5,8]]).T
N    = len(data)

# design matrix
X = np.ones([N,1])
# fit the model
b = np.linalg.solve(X.T@X,X.T@data)

# compare against the mean
m = np.mean(data)

# print the results
print(b,m)
```

```python
## many ways to compute the dot product

v1 = np.array([ 1, 2, 3, 4, 5 ])
v2 = np.array([ 0, -4,  -3, 6, 5 ])

# method 1
dp1 = sum( np.multiply(v1,v2) )

# method 2
dp2 = np.dot( v1,v2 )

# method 3
dp3 = np.matmul( v1,v2 )

# method 4
dp4 = 0  # initialize

# loop over elements
for i in range(len(v1)):

    # multiply corresponding element and sum
    dp4 = dp4 + v1[i]*v2[i]


print(dp1,dp2,dp3,dp4)
```

# VIDEO: Dot product properties: associative and distributive

```python
## Distributive property

# create random vectors
n = 10
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

# the two results
res1 = np.dot( a , (b+c) )
res2 = np.dot(a,b) + np.dot(a,c)

# compare them
print([ res1,res2 ])
```

```python
## Associative property

# create random vectors
n = 5
a = np.random.randn(n)
b = np.random.randn(n)
c = np.random.randn(n)

# the two results
res1 = np.dot( a , np.dot(b,c) )
res2 = np.dot( np.dot(a,b) , c )

# compare them
print(res1)
print(res2)


### special cases where associative property works!
# 1) one vector is the zeros vector
# 2) a==b==c
```

# VIDEO: The dot product from a geometric perspective
---


```python

# two vectors
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])

# compute the angle (radians) between two vectors
ang = np.arccos(np.clip(np.dot(v1,v2) / (np.linalg.norm(v1)*np.linalg.norm(v2)), -1, 1))


# draw them
fig = plt.figure()
# ax = fig.add_subplot(projection='3d')
ax = plt.axes(projection = '3d')
ax.plot([0, v1[0]],[0, v1[1]],[0, v1[2]],'b')
ax.plot([0, v2[0]],[0, v2[1]],[0, v2[2]],'r')

ax.set(xlim=(-6, 6), ylim=(-6, 6), zlim=(-6, 6))
plt.title('Angle between vectors: %s rad.' %ang)
plt.show()
```

```python
## equivalence of algebraic and geometric dot product formulas

# two vectors
v1 = np.array([ 2,  4, -3 ])
v2 = np.array([ 0, -3, -3 ])


# algebraic
dp_a = np.dot( v1,v2 )

# geometric
dp_g = np.linalg.norm(v1)*np.linalg.norm(v2)*np.cos(ang)

# print dot product to command
print(dp_a)
print(dp_g)
```

# VIDEO: Vector cross product
---


```python
# create vectors
v1  = [ -3,  2, 5 ]
v2  = [  4, -3, 0 ]

# Python's cross-product function
v3a = np.cross( v1,v2 )

# "manual" method
v3b = [ [v1[1]*v2[2] - v1[2]*v2[1]],
        [v1[2]*v2[0] - v1[0]*v2[2]],
        [v1[0]*v2[1] - v1[1]*v2[0]] ]

print(v3a,v3b)


fig = plt.figure()
# ax = fig.add_subplot(projection='3d')
ax = plt.axes(projection = '3d')

# draw plane defined by span of v1 and v2
xx, yy = np.meshgrid(np.linspace(-10,10,10),np.linspace(-10,10,10))
z1 = (-v3a[0]*xx - v3a[1]*yy)/v3a[2]
ax.plot_surface(xx,yy,z1,alpha=.2)

## plot the two vectors
ax.plot([0, v1[0]],[0, v1[1]],[0, v1[2]],'k')
ax.plot([0, v2[0]],[0, v2[1]],[0, v2[2]],'k')
ax.plot([0, v3a[0]],[0, v3a[1]],[0, v3a[2]],'r')


ax.view_init(azim=150,elev=45)
plt.show()
```

```python
# create a complex number
z = 3 + 4j

# magnitude
print( np.linalg.norm(z) )

# by transpose?
print( np.transpose(z)*z )

# by Hermitian transpose
print( np.transpose(z.conjugate())*z )


# complex vector
v = np.array( [ 3, 4j, 5+2j, complex(2,-5) ] )
print( v.T )
print( np.transpose(v) )
print( np.transpose(v.conjugate()) )
```
