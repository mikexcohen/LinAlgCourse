```python
import numpy as np
from sympy import *
import matplotlib.pyplot as plt
```

```python
# A tall matrix has a left inverse; its transpose has a right inverse.
m, n = 6, 3
A = np.random.randn(m, n)
AtA = A.T @ A
AAt = A @ A.T
print('Shape/rank of A^T A:', AtA.shape, np.linalg.matrix_rank(AtA))
print('Shape/rank of A A^T:', AAt.shape, np.linalg.matrix_rank(AAt))

Aleft = np.linalg.solve(AtA, A.T)
I_left = Aleft @ A
# A A^T has rank n < m, so it cannot be inverted.
# Use the wide matrix B for the right-inverse example instead.
B = A.T
Aright = np.linalg.solve(B @ B.T, B).T
I_right = B @ Aright
print('Left inverse check (Aleft @ A):', I_left)
print('Right inverse check (B @ Aright):', I_right)
assert np.allclose(I_left, np.eye(n))
assert np.allclose(I_right, np.eye(n))

AtA_inv = np.linalg.inv(AtA)
I_AtA = AtA_inv @ AtA
# A pseudoinverse of the singular Gram matrix gives a projector, not I.
AAt_inv = np.linalg.pinv(AAt)
I_AAt = AAt_inv @ AAt
print('Inverse check for A^T A:', I_AtA)
print('Projector for singular A A^T:', I_AAt)
```

```python
## show images
plt.subplot(331)
plt.imshow(A), plt.axis('off')
plt.title('A')

plt.subplot(332)
plt.imshow(AtA), plt.axis('off')
plt.title('A$^T$A')

plt.subplot(333)
plt.imshow(AAt), plt.axis('off')
plt.title('AA$^T$')

plt.subplot(335)
plt.imshow(Aleft), plt.axis('off')
plt.title('$(A^TA)^{-1}A^T$')

plt.subplot(336)
plt.imshow(Aright), plt.axis('off')
plt.title('$B^T(BB^T)^{-1}$, B=A^T')

plt.subplot(338)
plt.imshow(I_left), plt.axis('off')
plt.title('[ $(A^TA)^{-1}A^T ]$  A')

plt.subplot(339)
plt.imshow(I_right), plt.axis('off')
plt.title('B @ right inverse of B')

plt.show()
```
