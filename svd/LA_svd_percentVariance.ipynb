```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.signal
```

```python
# matrix sizes
m = 40
n = 30

# define a 2D Gaussian for smoothing
k = int( (m+n)/4 )
xx = np.linspace(-3,3,k)
[X,Y] = np.meshgrid(xx,xx)
g2d = np.exp(-(X**2 + Y**2) / (k/8))

# matrix
A = scipy.signal.convolve2d(np.random.randn(m,n),g2d,'same')
```

```python
Ascaled = A*1
# Center each feature before interpreting squared singular values as variance.
Ascaled = Ascaled - np.mean(Ascaled, axis=0, keepdims=True)

# SVD
U,s,V = np.linalg.svd(Ascaled)

# convert to percent variance
spct = 100*s**2/np.sum(s**2)  # variance is proportional to squared singular values

# plot the singular values for comparison
fig = plt.subplots(1,figsize=(7,5))

plt.subplot(211)
plt.plot(s,'ks-')
plt.xlabel('Component number')
plt.ylabel(r'$\sigma$')
plt.title('Raw singular values')

plt.subplot(212)
plt.plot(spct,'ks-')
plt.xlabel('Component number')
plt.ylabel('Variance explained (%)')
plt.title('Percent variance explained')

plt.tight_layout()
plt.show()
```
