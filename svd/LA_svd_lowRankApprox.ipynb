```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Construct the smoothed example matrix in this notebook as well.
from scipy.signal import convolve2d
m, n = 40, 30
k = int((m+n)/4)
xx = np.linspace(-3, 3, k)
X, Y = np.meshgrid(xx, xx)
g2d = np.exp(-(X**2 + Y**2)/(k/8))
A = convolve2d(np.random.randn(m, n), g2d, mode='same')
U, S, V = np.linalg.svd(A)

# number of components (singular "layers") to keep
nComps = 5

# reduced matrices
Ur = U[:,0:nComps]
Sr = S[0:nComps]
Vr = V[0:nComps,:]

# low-rank apporoximation
reconImage = Ur@np.diag(Sr)@Vr

# rank (confirm same as nComps)
print( 'rank =',np.linalg.matrix_rank(reconImage) )
```

```python
# squared reconstruction error
errormap = (reconImage-A)**2

# show the results!
plt.subplot(131)
plt.imshow(A)
plt.axis('off')
plt.title('Original')

plt.subplot(132)
plt.imshow(reconImage)
plt.axis('off')
plt.title('Low-rank')

plt.subplot(133)
plt.imshow(errormap, vmin=0)
plt.colorbar(label='Squared error')
plt.axis('off')
plt.title('error')
plt.show()
```
