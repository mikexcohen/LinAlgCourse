```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# some different matrices
S = np.zeros((5,), dtype=object)
S[0] = [ [ 4, 4], [ 4, 9] ]
S[1] = [ [-4,-1], [-3,-5] ]
S[2] = [ [ 0, 1], [ 2, 0] ]
S[3] = [ [ 1, 1], [ 1, 1] ]
S[4] = [ [-1,-2], [-3,-6] ]



# range for vector w
n = 30
wr = 2
wRange = np.linspace(-wr,wr,n)

# initialize quadratic form matrix
qf = np.zeros( (n,n) )


fig = plt.subplots(1,figsize=(8,8))

for i in range(5):
    
    # compute QF
    for xi in range(n):
        for yi in range(n):
            # this w
            w = np.transpose([ wRange[xi], wRange[yi] ])
            
            # QF
            qf[xi,yi] = w.T@S[i]@w
    
    # show the map
    plt.subplot(2,3,i+1)
    plt.imshow(qf.T,origin="lower",extent=[-wr,wr,-wr,wr])
    
    ## compute the matrix's definiteness based on the eigenvalues
    
    # x^T S x depends only on the symmetric part of a real matrix.
    symmetric_part = (np.asarray(S[i]) + np.asarray(S[i]).T)/2
    evals = np.linalg.eigvalsh(symmetric_part)
    tol = 1e-10 * max(1.0, np.linalg.norm(symmetric_part, 2))
    if np.all(evals > tol):
        defcat = 'Pos. def.'
    elif np.all(evals < -tol):
        defcat = 'Neg. def.'
    elif np.all(np.abs(evals) <= tol):
        defcat = 'Zero form'
    elif np.all(evals >= -tol):
        defcat = 'Pos. semidef.'
    elif np.all(evals <= tol):
        defcat = 'Neg. semidef.'
    else:
        defcat = 'Indefinite'

    # add title
    plt.title(defcat)

plt.show()
```
